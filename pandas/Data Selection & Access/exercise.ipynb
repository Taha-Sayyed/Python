```python
#Display the first 10 rows using iloc[].
df.iloc[:10,0:5]
```

```python
#Display rows with index labels 20 to 35 using loc[].
df.loc[20:35,['first_name','age','department']]
```

```python
#Using iloc[], retrieve rows 5 to 15 and only columns 0 to 4.
df.iloc[5:16,0:5]
```

```python
#Find count of all employees whose age is greater than 40.
len(df.loc[df['age']>40])
```

```python
#Find employees who belong to the Engineering department.
df.loc[df['department']=="Engineering"]
```

```python
#Find employees whose monthly salary is greater than ₹80,000 and whose experience is greater than 5 years.
df.loc[
    (df['monthly_salary_inr']>80000) & 
    (df['years_of_experience']>5)
]
```

```python
#Find employees who belong to either Mumbai or Pune.
df.loc[
    (df['city']=="Mumbai") |
    (df['city']=="Pune")
]
```

Find employees who:
- Have a performance score of at least 8.
- Have completed more than 10 projects.
- Are currently Active

```python
df.loc[
    (df['performance_score'].astype(int)>=8) &
    (df['projects_completed'].astype(int)>10) &
    (df['employment_status']=="Active")
]
```

```python
#Change the department of employee EMP1001 to Finance using loc[]
data=df.loc[df['employee_id']=="EMP1001"]
data['department']="Finance"
```

```python
#Update the work_from_home_days value to 5 for employees whose employment type is Full-time and whose city is Mumbai.
data=df.loc[
    (df['employment_type']=="Full-time") &
    (df['city']=="Mumbai")
]
data['work_from_home_days']=5
```
