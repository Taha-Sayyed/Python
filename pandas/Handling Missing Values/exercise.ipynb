Use isnull().sum() to calculate the number of missing values in every column. Identify:
- The column with the highest number of missing values.
- The column with the lowest number of missing values, excluding columns with zero missing values.
- The total number of missing cells in the DataFrame.

```python
max=-1
min=10000000
missing_value_sum=0
for i in range(0,14):
    data=df.iloc[:,i].isnull()
    print(f"Number of missing value in column {df.columns[i]} is {data.sum()}")
    missing_value_sum+=data.sum()
    if data.sum()>max:
        max=data.sum()
    if data.sum()<min:
        min=data.sum()
print(f"Max missing value count in dataframe: {max}")
print(f"Min missing value count in dataframe: {min}")
print(f"Sum of all missing value count in dataframe: {missing_value_sum}")
```
