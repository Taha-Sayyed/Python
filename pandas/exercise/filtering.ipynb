```python
import pandas as pd
```

```python
df=pd.read_csv('../Dataset/career_choices_2026.csv')
```

```python
# Find students that are expecting more then 20 LPA
result=df.loc[df['Expected_Salary_LPA']>20].head(5)
print(result[['Student_ID','Preferred_Career_2026','Expected_Salary_LPA']])
```

```python
# Calculate the overall average of the Expected salary in LPA
print(df['Expected_Salary_LPA'].mean())
```
