# Slicing using loc

loc is used for selecting specific subset of Data

- df.loc[row_condition, column_selection]
- DataFrame.loc[row_labels, column_labels]

```python
# Select a Single Value Using loc[]
print(df.loc[0,["Preferred_Career_2026"]])
```

```python
# Select Multiple Rows and Columns
#Note: WE can provide the label for accessing row
#Example: df.loc['Row_1':'Row_2',["Preferred_Career_2026","Primary_Skill"]]
print(df.loc[:5,["Preferred_Career_2026","Primary_Skill"]])
```

```python
# Select Rows Using a Condition

print(df.loc[df['Expected_Salary_LPA']>10].head())
```

```python
print(df.loc[df['Expected_Salary_LPA'] > 10, ['Student_ID','Expected_Salary_LPA']].head(5))
```

# Slicing using iloc

iloc is used for selecting specific subset of Data

- Syntax: pandas.DataFrame.iloc[] 
- Parameters:  Index position of rows in integer or list of integer. 
- Return type: Data frame or Series depending on parameters

```python
# Extracting row
print(df.iloc[3])
```

```python
# Selecting a particular value
print(df.iloc[3,5])
```

```python
# Selecting multiple rows and multiple column
print(df.iloc[[3,4,5],:2])
```
