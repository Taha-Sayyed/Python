# Pivot table

the biggest difference is that pivot() does not perform aggregation, whereas pivot_table() can perform aggregation.

DataFrame stores data in tabular format, whereas a Pivot Table summarizes and reorganizes data to extract insights

pivot() cannot be used in case of duplicate index-column combination. In that case, you have to use pivot_table()

```python
df=pd.read_csv("../Dataset/weather2.csv")
```

```python
df
```

```python
df.pivot_table(index='city',columns='date')
# Value inside the table is the average value
```

# Margins

In pd.pivot_table(), the margins parameter is used to calculate totals or subtotals (margins) across rows and columns.

```python
result=df.pivot_table(index="city",columns="date", margins=True,aggfunc=np.sum)
```

```python
print(result)
```

```python
result['humidity']
```
