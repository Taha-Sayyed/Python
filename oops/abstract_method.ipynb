```python
from abc import ABC, abstractmethod
```

```python
class Parent(ABC):
    def meth1(self):
        print("Parent Meth-1")

    @abstractmethod
    def meth2(self):
        pass
```

```python
class Child(Parent):
    def meth2(self):
        print("Parent Meth-2")
```

```python
c=Child()
```

```python
c.meth2()
c.meth1()
```
