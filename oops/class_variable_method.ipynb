```python
class Rectangle:

    #class variable
    count=0

    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
        Rectangle.count=Rectangle.count+1

    def area(self):
        return self.length*self.breadth

    #class method

    @classmethod
    def get_count(cls):
        # print(self.length) #Not error
        return cls.count

    
        
```

```python
rect1=Rectangle(10,10)
print(Rectangle.count)
print(rect1.get_count())
print(rect1.count)
```
