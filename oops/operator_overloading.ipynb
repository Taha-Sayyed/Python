Operator overloading is a form of compile-time polymorphism where the same operator behaves differently depending on the data type or objects on which it operates.

```python
a=10
b=20
print(a+b)
```

```python
c="Taha"
d="Ashraf"
print(c+d)
```

```python
class Student:
    def __init__(self, marks):
        self.marks = marks

    #Python calls this method internally when you use the + operator with objects.
    def __add__(self, other):
        return self.marks + other.marks


s1 = Student(80)
s2 = Student(90)
```

```python
print(s1+s2)
```

```python
s3=Student(100)
```

```python
print(s1+s2+s3)
```
