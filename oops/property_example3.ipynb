Python performed name mangling since we use double underscore

```python
class Person:
    def __init__(self,first_name,last_name):
        self.__first_name=first_name
        self.__last_name=last_name

    @property
    def name(self):
        return f"{self.__first_name} {self.__last_name}"
```

```python
p=Person("Taha","Sayyed")
```

```python
print(p.name)
```
