```python
class UserDefinedException(Exception):
    def __init__(self):
        self.msg="Negative dimension"
    def __str__(self):
        return self.msg
```

```python
class Area:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth

    def area(self):
        if self.length < 0 or self.breadth < 0 :
            raise UserDefinedException()
        else:
            return self.length*self.breadth
```

```python
try:
    a=Area(10,-5)
    result=a.area()
except UserDefinedException as e:
    print(e)
else:
    print(result)
```
