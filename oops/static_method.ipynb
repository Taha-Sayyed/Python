```python
class Rectangle:

    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth

    @staticmethod
    def calcArea(length,breadth):
        return length*breadth
```

```python
rect1=Rectangle(10,20)
print(rect1.calcArea(100,230))
print(Rectangle.calcArea(500,600))
```
