In Python, method overloading does not support with different number of argument. It supports method overloading with different type of argument 

```python
class Calculator:
    def add(self,a,b): #This will get shadowed
        return a+b

    def add(self,a,b,c):
        return a+b+c
```

```python
obj=Calculator()
```

```python
obj.add(10,20)
```

```python
obj.add(10,20,30)
```
