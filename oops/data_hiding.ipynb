Here the below code demonstrate the use of private variable and how child class use it

```python
class Parent:
    def __init__(self,d):
        self.__data=d

    def show(self):
        print(self.__data)
```

```python
p=Parent("Taha")
```

```python
p.__data #Since it the instance variable is used with double score
```

```python
p.show()
```

```python
class Child(Parent):
    def __init__(self, d):
        super().__init__(d)
    def display(self):
        print(self.__data)
```

```python
c=Child("Taha")
```

```python
c.show()
```

```python
c.display()
```

```python
p._Parent__data
```

```python
c._Parent__data
```
