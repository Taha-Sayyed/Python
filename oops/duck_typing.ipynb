```python
class Men:
    def talk(self):
        print("Men Talking")
    def walk(self):
        print("Men walking")
```

```python
class Dog:
    def walk(self):
        print("Dog walking")
```

```python
def person(pet):
    pet.walk()
    if hasattr(pet,'talk'):
        pet.talk()
```

```python
dog=Dog()
```

```python
person(dog)
```
