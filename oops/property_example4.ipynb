```python
class Person:
    def __init__(self,first_name,last_name):
        self._first_name=first_name
        self._last_name=last_name

    @property
    def name(self):
        return f"{self._first_name} {self._last_name}"
```

```python
p=Person("Taha","Sayyed")
```

```python
p.name
```
