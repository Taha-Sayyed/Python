```python
class Outer:

    def __init__(self):
        self.in_object=self.Inner()

    def display_inner_class_detail(self):
        print(self.in_object.show())
            

    class Inner:
        def __init__(self):
            self.inner_data="Inner Data"

        def show(self):
            return self.inner_data
```

```python
o=Outer()
```

```python
o.display_inner_class_detail()
```
