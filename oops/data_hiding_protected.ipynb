Here the below code demonstrate the use of protected variable and how child class use it

```python
class Parent:
    def __init__(self,d):
        self._data=d

    def show(self):
        print(self._data)
```

```python

p=Parent("Taha")
```

```python
p._data
```

```python
p.show()
```

```python
class Child(Parent):
    def __init__(self, d):
        super().__init__(d)
    def display(self):
        print(self._data)
```

```python
c=Child("Taha")
```

```python
c.show()
```

```python
c.display()
```
