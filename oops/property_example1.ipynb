```python
class Rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth

    @property
    def length(self):
        return self._length

    @length.setter
    def length(self,len):
        if(len<0):
            print("Not possible")
        else:
            self._length=len
        
```

```python
rect=Rectangle(10,20)
print(rect.length)
rect.length=20
print(rect.length)
```
